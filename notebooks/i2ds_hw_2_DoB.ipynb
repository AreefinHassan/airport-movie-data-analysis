
# Use the date to calculate the age values in years, pick the ones ranging from 1 to 99 years. **bold text**

```python
# calling libraries

from datetime import datetime
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import math
import scipy.stats as stats
```

```python
with open('/content/drive/MyDrive/i2ds/hw_2/hw_2_2/DoB.txt', 'r') as f:
    dob = f.read()

# pre-processing the data
     
dob_list = dob.split('\n')
# print(dob_list)
curr_date = datetime.now()
# print(curr_date)
curr_year = curr_date.year
# print(curr_year)

# ages_list=[]
ages = []

# rounding birth years of the dataset:
for dates in range(len(dob_list)):
    date_mm_yy = dob_list[dates].split('-')
    if len(date_mm_yy) == 3:
        year = date_mm_yy[2] 
    if len(year) == 2:
        year ='19' + year  #19XX years
    ages.append(curr_year - int(year))

# print(dob_list)
age_dic = {'Birth_dates': dob_list, 'Ages': ages}
df_ages_all = pd.DataFrame(age_dic) 
df_ages_all
```

```python
# ranging ages from 1 to 99 as per discord instruction
df_ages_all_1_99 = df_ages_all[df_ages_all['Ages'].between(1, 99, inclusive=True)]
```

```python
# verifying age range:
print("Min Age:", df_ages_all_1_99["Ages"].min())
print("Max Age", df_ages_all_1_99["Ages"].max())
```

```python
# seperating the 1st and Last digit:

x = df_ages_all_1_99['Ages']

# % -> modulus
# // -> divides with integral result (discard remainder)

df_ages_all_1_99['first_digit'] = (x % 100) // 10  
df_ages_all_1_99['last_digit'] = (x % 10)
df_ages_all_1_99
```

# **distribution of the age**

```python
plt.hist(df_ages_all_1_99['Ages'], bins=100, color="red")
plt.legend(labels=["Age"])
plt.title("Origial visualization From the given Data")
plt.xlabel("Age")
plt.ylabel("Count")
```
